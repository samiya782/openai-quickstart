```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import os

load_dotenv()
if not os.getenv("GOOGLE_API_KEY"):
    raise ValueError("GEMINI_API_KEY not found in .env file or environment variables.")
print("Gemini API Key loaded successfully.")

python_prompt = PromptTemplate.from_template("""
你是一名经验丰富的Python开发者。请根据以下要求编写Python代码：
{input}
""")
java_prompt = PromptTemplate.from_template("""
你是一名经验丰富的Java开发者。请根据以下要求编写Java代码：
{input}
""")
llm = ChatGoogleGenerativeAI(model="gemini-2.5-pro")

chain = (
    { "input": RunnablePassthrough() }
    | {
        "python_code": python_prompt | llm | StrOutputParser(),
        "java_code": java_prompt | llm | StrOutputParser()
    }
    | RunnableLambda(lambda x: f"Python代码:\n{x['python_code']}\n\nJava代码:\n{x['java_code']}")
)
chain.invoke("编写一个计算斐波那契数列的函数")
```
